## Cuide dos valores ausentes (se houver algum)

Os valores ausentes encontrados serão preenchidos pela mediana da sua coluna e para alguns que possuem colunas relacionadas as que possuem valores ausentes e pode-se criar categorias e substituir os valores ausentes pela mediana de cada categoria.

__Preenchendo os valores ausentes em model year__

Preenchendo os valores ausentes utilizando a mediana da coluna model_year onde ha valores ausentes.

```python
median_model_year = df_vehicles['model_year'].median()
df_vehicles['model_year'] = df_vehicles['model_year'].fillna(median_model_year)
```

__Preenchendo os valores ausentes em cylinders__

Nesta celula abaixo irei criar um dicionario que ira armazenar as medianas de cada tipo de veiculo e esta mediana armazenada ira ser utilizada para preencher valores ausentes em cylinders que serao preenchidos baseado na mediana do seu tipo isso sera interessante porque a coluna type nao contem valores ausentes entao todas as entradas poderao ser referenciadas pelo seu tipo.

```python
tipos = ['SUV','truck','sedan','pickup','coupe','wagon','mini-van','hatchback','van','convertible','other','offroad','bus'   ]
median_tipos = {}
for tipo in tipos:
    median_tipos[tipo] = df_vehicles.query(f'type == "{tipo}"')['cylinders'].median()
print(median_tipos)

```

Nesta situação irei usar uma função customizada para preencher os valores medianos dos cilindros agregados pelo tipo de veiculo.

```python

def fill_na_cylinders(row):
    cylinder = row['cylinders']
    type_vehicle = row['type']
    if math.isnan(cylinder) == True:
        return int(median_tipos[type_vehicle])
    else:
        return int(cylinder)


df_vehicles['cylinders'] = df_vehicles.apply(fill_na_cylinders, axis=1)
```

__Preenchendo os valores ausentes em odometer__

No caso do campo odometer apenas preencher os valores ausentes com o metodo fillna atraves da mediana da coluna odometer ira eliminar o problema.

```python
median_odometer = df_vehicles['odometer'].median()
df_vehicles['odometer'] = df_vehicles['odometer'].fillna(value=median_odometer)

df_vehicles.info()
```

__Preenchendo os valores ausentes em paint color__

Os valores ausentes em paint_color irao ser preenchidos com o valor da moda, ou o valor que mais aparece na coluna paint_color.

```python
mode_paint_color = df_vehicles['paint_color'].mode()
df_vehicles['paint_color'] = df_vehicles['paint_color'].fillna(value=mode_paint_color[0])
df_vehicles.info()
```

__Preenchendo os valores ausentes em is 4wd__

Para a coluna is_4wd pude investigar alguns carros na internet onde os valores para esta coluna estao ausentes e pude perceber que se trata de veiculos que nao sao 4wd por alguma razao nao se utilizou zero como falso mas irei definir para zero assim ficara mais uniforme e coerente.

```python
def fill_is_4wd(row):
    value = row['is_4wd']
    if math.isnan(value) == True:
        return int(0)
    else:
        return int(value)

df_vehicles['is_4wd'] = df_vehicles.apply(fill_is_4wd, axis=1)
df_vehicles.info()
```

## Corrija os tipos de dados

__Convertendo os valores da coluna date posted de string para datetime__

Os dados na coluna date_posted estao no tipo de dados string e precisam ser convertidos para o tipo datetime para fazer isso irei criar uma funcao customizada e aplicar a funcao a toda a coluna date_posted.

```python
def convert_to_datetime(row):
    date = row['date_posted']
    return pd.to_datetime(date, format='%Y-%m-%d')
df_vehicles['date_posted'] = df_vehicles.apply(convert_to_datetime, axis=1)
```

__Convertendo os valores da coluna model year de float para inteiro__


Os dados na coluna model year estao registados no tipo de dado float mas o tipo de dados mais apropriado para se guardar dados como ano e o int entao os dados serao convertidos para int.


```python
def float_to_int(row):
    value = row['model_year']
    return int(value)
df_vehicles['model_year'] = df_vehicles.apply(float_to_int, axis=1)
```

__Convertendo os valores da coluna odometer de float para inteiro__

Os dados na coluna odometer tambem precisam ser convertidos para o tipo de dados inteiro porque a parte de ponto flutuante nao esta a ser realmente utilizada entao ficaria melhor no tipo de dado int.

```python
def odometer_to_int(row):
    value = row['odometer']
    return int(value)
df_vehicles['odometer'] = df_vehicles.apply(odometer_to_int, axis=1)
```

## Enriqueça os dados

__Criando colunas as colunas Mes, ano e dia da semana__

Criando as colunas mes, ano e dia da semana que foi criada a propaganda atraves da coluna date_posted desta forma sera mais facil fazer analise usando apenas estes valores isolados.

```python
# Adicione valores data/hora para quando a propaganda estiver colocada
df_vehicles['month_posted'] = df_vehicles['date_posted'].dt.month

df_vehicles['year_posted'] = df_vehicles['date_posted'].dt.year

df_vehicles['weekday_posted'] = df_vehicles['date_posted'].dt.weekday

df_vehicles.head(10)
```

__Criando uma coluna que ira armazazenar a idade do veiculo__

Esta coluna ira conter a idade do veiculo na subtraindo da data em que foi publicada a propaganda o model_year.

```python
# Adicione a idade do veículo quando a propaganda for colocada
df_vehicles['vehicle_age'] = df_vehicles['year_posted'] - df_vehicles['model_year']
```

__Calculando a quilometragem media por ano__

Na celula abaixo e criada uma nova coluna que armazena a media de kilometragem de cada veiculo por ano.


```python
# Adicione a quilometragem média por ano

df_vehicles['avg_km_year'] =  df_vehicles['odometer'] / (df_vehicles['vehicle_age'] + 1)
```

__Substituir os valores de condition por uma escala numerica__

Na celula abaixo os valores da coluna condition sao substituidos por uma escala numerica onde:
- new = 5
- like new = 4
- excellent = 3
- good = 2
- fair  = 1
- salvage = 0

```python
# Isso deve ajudar a substituir os valores condicionais por alguma coisa que pode ser manipulada mais facilmente

valores = {'new':5,
'like new' : 4,
'excellent' : 3,
'good' : 2,
'fair' : 1,
'salvage' : 0}

def create_scale(row):
    condition = row['condition']
    return valores[condition]

df_vehicles['condition'] = df_vehicles.apply(create_scale, axis=1)

```

## Estude e cuide dos valores atípicos



__Calculando os limites inferiores para as colunas dos parametros definidos__

Nesta secção é criada uma função que irá filtrar os outliers e retornar os valores ja filtrados.

```python
# Criando funcao para remover os outliers e retornar o dataframe sem os outliers

def remove_outliers(column):
    param = column
    q1 = df_vehicles[column].quantile(.25)
    q3 = df_vehicles[column].quantile(.75)
    iqr = q3 - q1
    lim_inf = q1 - (1.5*iqr)
    lim_sup = q3 + (1.5*iqr)
    df_filtered = df_vehicles.query(column +'> @lim_inf and '+column+' < @lim_sup')
    return df_filtered
```

Criando um dataframe para a coluna __price__ com os seus outliers filtrados.

```python
# Dataframe sem os outliers de price 
price_nonoutliers = remove_outliers('price')
price_nonoutliers.hist(column='price')
```

Criando um dataframe para a coluna __vehicle age__ com os seus outliers filtrados.

```python
# Dataframe sem os outliers de vehicle_age 

age_nonoutliers = remove_outliers('vehicle_age')
age_nonoutliers.hist(column='vehicle_age')
```

Criando um dataframe para a coluna __odometer__ com os seus outliers filtrados.

```python
# Dataframe sem os outliers de odometer 

odometer_nonoutliers = remove_outliers('odometer')
odometer_nonoutliers.hist(column='odometer')
```

Criando um dataframe para a coluna __cylinders__ com os seus outliers filtrados que praticamente é o mesmo grafico pois para esta coluna não há outliers.

```python
# Dataframe sem os outliers de cylinders 

cylinders_nonoutliers = remove_outliers('cylinders')
cylinders_nonoutliers.hist(column='cylinders')
```

Criando um dataframe para a coluna __condition__ com os seus outliers filtrados tal como para a coluna cylinders a coluna condition apresenta praticamente o mesmo grafico.

```python
# Dataframe sem os outliers de condition 

condition_nonoutliers = remove_outliers('condition')
condition_nonoutliers.hist(column='condition')
```

## Fatores de preço

O preço de um veiculo depende de fatores como a __idade__, __Quilometragem__, __Condição__, __Transmissão__ e __cor__ do veiculo, é possivel se chegar a esta informação analizando os graficos gerados abaixo que ilustram oque acontece ao preço do veiculo a medida que o grafico avança no eixo x, o preço aumenta ou diminui a medida que o eixo avança.

__Criando um dataframe sem qualquer valor outlier__

```python
def remove_outliers2(df,column):
    param = column
    q1 = df[column].quantile(.25)
    q3 = df[column].quantile(.75)
    iqr = q3 - q1
    lim_inf = q1 - (1.5*iqr)
    lim_sup = q3 + (1.5*iqr)
    df_filtered = df.query(column +'> @lim_inf and '+column+' < @lim_sup')
    return df_filtered

df_filtered = remove_outliers2(df_vehicles, 'price')
df_filtered = remove_outliers2(df_filtered, 'vehicle_age')
df_filtered = remove_outliers2(df_filtered, 'odometer')
df_filtered = remove_outliers2(df_filtered, 'condition')

```

## Conclusão geral

Nesta analise foi possivel se estudar como o preço de um veiculo pode variar e de que depende o mesmo entre os outros dados fornecidos acerca dos veiculos e propagandas apresentadas. A analise dos dados indicou que os seguintes fatores influenciam os preços dos veiculos:

- __Transmissões:__ os veiculos com transmissão automatica tendem a ter o preço mais alto pois como mediana o preço esta por volta dos 1200 e de seguida vêm as trasmissões manuais que têm os preços mais baixos em media dos 1000.

- __Cor:__ As cores amarela, laranja e preta são as dos veiculos que normalmente têm preços mais altos e as cores verde e castanha são as dos veiculos com preços mais baixos.

- __Condições:__ em geral veiculos em condições razoaveis têm os preços mais baixos e os veiculos em condições "bom", "excelente", e "como novo" são os que normalmente têm preços altos.

- __Quilometragem:__ o graficos de dispersão permite visualizar que o preço é inversamente proporcional a quilometragem do veiculo, em outras palavras, quanto menor a quilometragem maior o preço do veiculo e vice-versa.

- __Idade do Veículo:__ é possivel identificar nos gráficos que quanto menor a idade do veiculo, maior o preço e vice-versa, portanto são também inversamente proporcionais.

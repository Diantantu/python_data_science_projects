## 1 Inicialização

De forma a poder fazer a analise, preparação e treinamento de dados serão reutilizadas bibliotecas em python muito conhecidas para nos ajudar em tarefas que precisam ser executadas para podermos chegar a solução do problema.

__Carregando Bibliotecas Necessarias__

Na celula abaixo são importadas as bibliotecas que serão usadas no nosso estudo e estas são: pandas, numpy, math, pyplot, stats e seaborn.

```python
#Carregando as bibliotecas necessarias
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
from scipy import stats as st
import seaborn as sns

#Bibliotecas de manipulação de modelos e treinamento
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

## 2 Carregando os dados

Nesta celula serão importados os dados de fontes externas para se fazer a analise e calculos que no fim nos darão informações acerca do mercado de taxis de aplicativos. 

```python
#Carregando os dados para um objeto pandas dataframe

df = pd.read_csv("/datasets/users_behavior_upd.csv")
```

## 3 Divisão dos dados 

É necessario fazer a divisão dos dados em conjunto de treinamento, conjunto de validação e conjunto de teste para poder se avaliar o modelo que será criado com conjuntos de dados diferentes em cada etapa da criação do modelo e treinamento.

```python
random_s = 54321
train_df, test_df = train_test_split(df, test_size=0.2, random_state=random_s)

train_df, val_df = train_test_split(train_df, test_size=0.25, random_state=random_s)
```

## 4 Estudo do modelo que será usado

Iremos criar um modelo do tipo floresta aleatória e outro de regressão logistica para poder ser feita a comparação dos dois modelos e verificar qual modelo é mais eficiente para fazer classificações com os nossos conjuntos de dados.

__Separando as caracteristicas e o objetivo para treino e validação__

```python
features_train = train_df.drop(['is_ultra'], axis=1)
target_train = train_df['is_ultra']

features_valid = val_df.drop(['is_ultra'], axis=1)
target_valid = val_df['is_ultra']

best_score = 0
best_est = 0

```

### Modelo Floresta Aleatória

Aqui irei treinar um modelo de floresta aleatoria e ajustar os hiperparametros até encontrar o que funciona melhor neste problema e conjunto de dados

__Verificando qual é a quantidade de Arvores Decisão ideal para este Modelo__

```python
for est in range(1, 50):
    model = RandomForestClassifier(random_state=random_s, n_estimators= est)
    model.fit(features_train, target_train)
    score = model.score(features_valid, target_valid)
    
    if score > best_score:
        best_score = score
        best_est = est

        
print(f"Melhor score: {best_score}")
print(f"Sua quantidade de Estimators: {best_est}")
```

__Conclusões__

Após treinar o conjunto de dados e validar com o conjunto de dados foi possivel se chegar a 84% de eficacia com 49 arvores de decisão oque é a mais alta mesmo que se aumentar a quantidade de arvores a acuracia não melhora.

### Modelo Regressão Logística

Será usado o modelo de Regressão Logística e será treinado com o conjunto de dados de treino e depois vai ser avaliada a sua acuracia no  conjunto de dados de treino e no conjunto de dados de validação.

```python
model_lr = LogisticRegression(random_state=random_s, solver="liblinear")

model_lr.fit(features_train, target_train)

score_train_lr = model_lr.score(features_train, target_train)

score_valid_lr = model_lr.score(features_valid, target_valid)

print(f"Acuracia no conjunto de treinamento: {score_train_lr}")

print(f"Acuracia no conjunto de validacao: {score_valid_lr}")
```

__Breves Conclusões__

Foi possivel criar um modelo de regressão logistica e foi verificada a acuracia do modelo nos dados de treinamento e nos dados de validação oque foi muito baixo para os dois no conjunto de treinamento a acuracia foi de 71% e no conjunto de validação 73%. Pese embora a acuracia melhorou no conjunto de validação ainda assim é muito baixa, o nosso modelo ainda é muito impreciso pois só acerta em 73% dos casos.

#### Conclusão

Apos criação de modelo de floresta aleatoria e de regressão logistica foi possivel ser verificar que o modelo de floresta aleátoria é muito mais assertivo chegando a uma acuracia de 84% com 49 arvores de decisão, foi possivel perceber-se também que o modelo de regressão logistica foi muito mais rapido no treinamento do que o modelo de floresta aleatória mas as suas predições não foram muito boas.
